```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#1)Load the dataset:
url = "https://raw.githubusercontent.com/ArchanaInsights/Datasets/main/marketing_campaign.csv"
df = pd.read_csv(url)
```

```python
df.shape
```

```python
#2)Descriptive Analysis:

#Basic Structure:
#Print first few rows:
print(df.head())
```

```python
#Number of rows and columns:
print("Shape:", df.shape)
```

```python
#Concise summary of dataset:
df.info()
```

```python
#Descriptive Statistics:
df.describe()
```

```python
#Data Exploration:

#Unique Campaign_ID values:
print("Unique Campaign ID Values:", df['Campaign_ID'].nunique())
```

```python
#Unique values of Location and Customer_Segment columns:
print("Unique Locations:", df['Location'].unique())
print("Unique Customer Segments:", df['Customer_Segment'].unique())
```

```python
#Count occurrences of each category in Campaign_Type and Channel_Used:
print(df['Campaign_Type'].value_counts())
print(df['Channel_Used'].value_counts())
```

```python
#3)Exploratory Data Analysis (EDA) and Visualization:

#Campaign Performance:
#Scatter Plot to show relationship between Acquisition_Cost and ROI:
plt.figure()
sns.scatterplot(data=df, x='Acquisition_Cost', y='ROI')
plt.title("Relationship Between Acquisition Cost and ROI")
plt.show()
```

```python
#Bar chart to visualize average Conversion_Rate for different Channel_Used categorized by Campaign_Type:
plt.figure()
sns.barplot(data=df, x='Channel_Used', y='Conversion_Rate', hue='Campaign_Type', palette='dark:red')
plt.title("Average Conversion Rate by Channel Used & Campaign Type")
plt.show()
```

```python
#Box Plot to visualize distribution of Engagement_Score across different Campaign_Type:
plt.figure()
sns.boxplot(data=df, x='Campaign_Type', y='Engagement_Score', color="green")
plt.title("Distributin of Engagement Score by Campaign Type")
plt.xlabel("Campaign Type")
plt.ylabel("Engagement Score")
plt.show()
```

```python
#Analyse average ROI by Company using Bar Chart:
avg_roi = df.groupby('Company')['ROI'].mean().reset_index()

plt.figure(figsize=(10,5))
sns.barplot(x='Company', y='ROI', data=avg_roi)
plt.title("The Average ROI by Company")
plt.show()
```

```python
#Correlation between Engagement_Score and Conversion_Rate using heatmap:
corr = df[['Engagement_Score', 'Conversion_Rate']].corr()

plt.figure()
sns.heatmap(corr, annot=True, cmap='Set3')
plt.title("Correlation Between Engagement Score and Conversion Rate")
plt.show()
```

```python
#Customer segmentation:
#Count Plot to visualize distribution of Target_Audience:
sns.set(style="dark")
plt.figure()
sns.countplot(data=df, x='Target_Audience', hue='Target_Audience')
plt.title("Distribution Of Target Audience")
plt.xlabel("Target Audience")
plt.show()
```

```python
#Using Bar Chart for identifying which Customer_Segment has the highest Conversion_Rate for each Language:
segment_lang = df.groupby(['Language','Customer_Segment'])['Conversion_Rate'].mean().reset_index()

plt.figure()
sns.barplot(data=segment_lang, x='Language', y='Conversion_Rate', hue='Customer_Segment')
plt.ylabel("Conversion Rate")
plt.title("Conversion Rate by Customer Segment & Language")
plt.show()
```

```python
#Distribution of Aquisition Cost across Customer Segment categorized by Channel used using Box Plot:
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x='Customer_Segment', y='Acquisition_Cost', hue='Channel_Used', palette="Accent")
plt.xlabel("Customer Segment")
plt.ylabel("Aquisition Cost")
plt.title("Distribution of Acquisition Cost across Customer Segment using Box Plot")
plt.show()
```

```python
#Using Bar Chart to analyze average conversion rate by language:
plt.figure()
sns.barplot(data=df, x='Language', y='Conversion_Rate', color='Brown')
plt.ylabel("Conversion Rate")
plt.title("Average Conversion Rate by Language")
plt.show()
```

```python
#Channel Effectiveness:

#Using Bar Chart compare Engagement Score for different Channels Used:
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Channel_Used', y='Engagement_Score', hue='Campaign_Type', palette='viridis')
plt.title("Engagement Score by Channel & Campaign Type")
plt.xlabel("Channel Used")
plt.ylabel("Engagement Score")
plt.show()
```

```python
#Distribution of total ROI across different Channel Used using Pie Chart:
roi_channel = df.groupby('Channel_Used')['ROI'].sum()

plt.figure()
plt.pie(roi_channel, labels=roi_channel.index, autopct='%1.1f%%')
plt.title("Distribution of Total ROI by Channel")
plt.show()
```

```python
#Scatter Plot to show relationship between Clicks and Impressions for Channel Used:
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='Impressions', y='Clicks', hue='Channel_Used', marker="*")
plt.title("Relationship Between Impressions and Clicks")
plt.show()
```

```python
#Time Based Analysis:

#Distribution of Duration using Histogram:
plt.figure()
plt.hist(df['Duration'], color="pink")
plt.title("Distribution of Duration")
plt.show()
```

```python
#Analyse Conversion_Rate over Date by Company using Line Chart:
avg_conversion = df.groupby(["Date", "Company"])["Conversion_Rate"].mean().reset_index()
plt.figure(figsize=(15,5))
sns.lineplot(data=avg_conversion, x="Date", y="Conversion_Rate", hue="Company", linewidth=1, marker="*")
plt.title("Conversion Rate Over Date by Company")
plt.ylabel("Conversion Rate")
plt.legend(title="Company")
plt.show()
```

```python
#Examine Engagement Score over Date using Line Chart:
eng_average = df.groupby('Date')['Engagement_Score'].mean().reset_index()
plt.figure()
sns.lineplot(data=eng_average, x='Date', y='Engagement_Score', color="black", linewidth=1)
plt.title("Engagement Score Over Date")
plt.ylabel("Engagement Score")
plt.show()
```

```python
#Geographic Analysis:

#Using Bar Chart determine location with highest Aquisition Cost:
high_cost = df.groupby('Location')['Acquisition_Cost'].mean().reset_index()
plt.figure(figsize=(10,5))
sns.barplot(data=high_cost, x='Location', y='Acquisition_Cost', color="lightgreen")
plt.title("Location with Highest Acquisition Cost")
plt.ylabel("Acquisition Cost")
plt.show()
```

```python
#Using Bar Chart visualize Conversion Rate by different Location categorized by Target Audience:
conv = df.groupby(['Location','Target_Audience'])['Conversion_Rate'].mean().reset_index()
plt.figure(figsize=(10,5))
sns.barplot(data=conv, x='Location', y='Conversion_Rate', hue='Target_Audience', palette="Pastel2")
plt.title("Conversion Rate by Different Location & Audience")
plt.ylabel("Conversion Rate")
plt.show()
```

```python
#Using Pie Chart illustrate proportion of ROI by Location:
roi_loc = df.groupby('Location')['ROI'].sum()
plt.figure()
plt.pie(roi_loc, labels=roi_loc.index, autopct='%1.1f%%')
plt.title("Proportion of ROI by Location")
plt.show()
```

```python
import pandas as pd
import numpy as np

# Define column names for the games dataset
columns = [
    'Game_ID', 'Title', 'Genre', 'Platform', 'Release_Year',
    'Developer', 'Publisher', 'Rating', 'Price', 'Copies_Sold_Millions'
]

# Generate 10 rows of data
data = {
    'Game_ID': np.arange(1, 11),
    'Title': [f'Game {i}' for i in range(1, 11)],
    'Genre': np.random.choice(['Action', 'Adventure', 'RPG', 'Strategy', 'Sports'], 10),
    'Platform': np.random.choice(['PC', 'PlayStation', 'Xbox', 'Nintendo Switch'], 10),
    'Release_Year': np.random.randint(2010, 2024, 10),
    'Developer': [f'DevCo {i}' for i in range(1, 11)],
    'Publisher': [f'PubCorp {i}' for i in range(1, 11)],
    'Rating': np.random.uniform(2.5, 5.0, 10).round(1),
    'Price': np.random.uniform(19.99, 69.99, 10).round(2),
    'Copies_Sold_Millions': np.random.uniform(0.5, 20.0, 10).round(1)
}

# Create the DataFrame
games_df = pd.DataFrame(data, columns=columns)

# Display the DataFrame
print(games_df)
```

```python
print("Games categorized by Genre:\n", games_df['Genre'].value_counts())
```

```python
games_df['Rating'] = games_df['Rating'].astype(int)
games_df['Price'] = games_df['Price'].astype(int)
games_df['Copies_Sold_Millions'] = games_df['Copies_Sold_Millions'].astype(int)

print(games_df.head())
```

```python
print(games_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(data=games_df, x='Game_ID', y='Price', marker='*', s=100) # s for marker size
plt.title('Game ID vs. Price')
plt.xlabel('Game ID')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```
